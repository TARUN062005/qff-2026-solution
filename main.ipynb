## A/B table

Each row is one processor and one execution mode. Circuit columns repeat because compilation does not depend on the noise model. The CSV file is the source for the printed values.

```python
import csv

table_path = root / "results" / "tables" / "AB_comparison.csv"
with table_path.open(newline="", encoding="utf-8") as handle:
    rows = list(csv.DictReader(handle))

columns = [
    "processor",
    "execution_mode",
    "expected_cut",
    "expected_cut_standard_error",
    "approximation_ratio",
    "transpiled_depth",
    "transpiled_two_qubit_count",
    "explicit_swap_count",
    "routing_overhead_cx",
    "physical_qubits_allocated",
    "physical_qubits_used",
]
for row in rows:
    print(" | ".join(row[column] for column in columns))
```
